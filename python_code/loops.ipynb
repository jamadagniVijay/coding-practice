```python
# CODE-9: GCD (Greatest Common Divisor) - https://freedium-mirror.cfd/medium.com/@amitvsolutions/hello-recursion-68a9c2c19be4

"""
greatest common divider
32,6
32 = 2x2x2x2x2
6 = 3x2

32 % 6 = 2
2 %6 = 2

GCD = 2

32,14
32 = 2x2x2x2x2
14 = 7x2
GCD = 2

12/8

GCD using euclidian logic
The greatest common divisor of two numbers does not change if you replace the larger number with the remainder of their division
"""
def gcd(a,b):
    # base-code
    if a % b == 0:
        return b
    else :
        # recursion and accumulation code
        return gcd(b, a%b)

print(f"{gcd(6757645,5645)}")
```

```python
# CODE-10: Tower of Hanoi - https://freedium-mirror.cfd/medium.com/@amitvsolutions/hello-recursion-68a9c2c19be4
"""

Assumtions : 
    *   There will be only 1 of source, destination and Auxillary/Helper rod
    *   The standard recursive Tower of Hanoi algorithm is mathematically guaranteed
         to keep the disks in the correct sorted order at all times on every peg
        ⁂ to begin with too, they will be in sorted order
    * Total number of steps = 2^n - 1


[number_of_disks(n), source_rod(s), auxillary_rod(a), destination_rod(d)]
Towers of Hanoi as an Example of Recursion - http://youtube.com/watch?v=Ajy8XweC3L8
⭐Alogorithm parameters (always) : 
Recurssion begins with 
1️⃣ -> number_of_disks_to_be_moved, source_peg, intermediate_peg, destination_peg
    
Algorithm explained in words -> 

    🅱️Base Condition : if n=1 -> move disk from source to destination and done

    Else -> 
        n-1 disks will be moved from source to auxillary, keeping destination as intermediate
        move disk source -> destination
        n-1 disks from auxxillary to destination keeping source as intermediate

"""

def tower_of_hanoi(number_of_disks,source_peg,destination_peg,auxillary_peg):
    # base-check
    if number_of_disks == 1:
        return print(f"Move disk 1 from {source_peg} to {destination_peg}")

    tower_of_hanoi(number_of_disks-1,source_peg,auxillary_peg,destination_peg)

    print(f"Move disk {number_of_disks} from {source_peg} to {destination_peg}")

    tower_of_hanoi(number_of_disks-1,auxillary_peg,destination_peg,source_peg)

tower_of_hanoi(3,"Src","Aux","Dest")
    
```

### Claude-Assisted-DSA-training

```python
"""Claims reconciliation, the kind of thing that runs against a Redshift extract.
 You have a list of claim amounts and a target total, and you need to find the two claims that sum to it."""

def find_claim_pair(amounts, target):
    """
    amounts: list of ints, e.g. [340, 1250, 890, 410]
    target:  int, e.g. 1230
    Return the two indices whose amounts sum to target, e.g. (0, 3).
    Assume exactly one valid pair exists.
    """
    pairs = []
    target_complement = {}
    for idx,amount in enumerate(amounts):
        if(amount in target_complement):
            pairs.append((idx,target_complement[amount]))
        target_complement[(target - amount)] = idx
    return pairs

amounts = [340, 1250, 890, 410]
target = 1230
print(f"{find_claim_pair(amounts,target)}")
print("space complexity -> O(n) and Time complexity -> O(n)")
print("Edge case : if two different amounts produce the same complement key, the later one overwrites the earlier index.")
```

```python
"""Duplicate patient detection again, but harder — and this is a real problem in health data, 
where the same human gets registered twice with fields entered in different orders.

def group_possible_duplicates(patients):
    patients: list of dicts, each with 'id' and 'name' keys, e.g.
      [{"id": "P1", "name": "VIJAY KUMAR"},
       {"id": "P2", "name": "KUMAR VIJAY"},
       {"id": "P3", "name": "ANITA RAO"},
       {"id": "P4", "name": "RAO ANITA"},
       {"id": "P5", "name": "SURESH BABU"}]

    Two patients are possible duplicates if their names contain the same
    words, regardless of order. Return a list of lists of ids, one inner
    list per group. For the input above: [["P1","P2"], ["P3","P4"], ["P5"]]
"""
def group_possible_duplicates(patients):

    unique_patients = {}
    """canonicalization"""
    for patient in patients:
        fullName = patient["name"]
        parts = sorted(fullName.split())
        
        sortedName = " ".join(parts) 

        if sortedName in unique_patients:
            unique_patients[sortedName].append(patient["id"])
        else:
            unique_patients[sortedName] = []
            unique_patients[sortedName].append(patient["id"])
        
    return unique_patients

"""More efficient version"""
def group_possible_duplicates(patients):
    groups = {}
    for patient in patients:
        key = " ".join(sorted(patient["name"].split()))
        groups.setdefault(key, []).append(patient["id"])
    return list(groups.values())

patients = [{"id": "P1", "name": "VIJAY KUMAR"},
       {"id": "P2", "name": "KUMAR VIJAY"},
       {"id": "P3", "name": "ANITA RAO"},
       {"id": "P4", "name": "RAO ANITA"},
       {"id": "P5", "name": "SURESH BABU"}]

print(f"{group_possible_duplicates(patients)}")
print("The space complexity is O(n) and time compexity is O(n)")
"""Let n = number of patients, k = words per name:

Per patient: split O(k), sort O(k log k), join O(k) → dominated by O(k log k)
Across all patients: O(n · k log k) time, O(n · k) space.
With k tiny and bounded (names have 2–4 words), this behaves like O(n) — and saying 
"effectively linear since k is small" is a perfectly good interview answer. 
But you have to show you saw the k. Claiming flat O(n) reads as not having looked."""
```

```python
"""def find_conflicting_coverage(members):
    members: list of dicts, e.g.
      [{"member_id": "M1", "payer": "AETNA",  "plan": "PPO-GOLD"},
       {"member_id": "M2", "payer": "aetna",  "plan": "ppo-gold"},
       {"member_id": "M3", "payer": "CIGNA",  "plan": "HMO-BASE"}]

    Two members are on the same coverage if payer and plan match,
    ignoring case and surrounding whitespace.
    Return only the groups that contain more than one member.
"""
def find_conflicting_coverage(members):
    patients_on_same_coverage = {}

    for member in members:
        # ambiguious: key = f"{member["payer"].upper()}-{member["plan"].upper()}"
        # key should not create any ambiguity
        # alternative 1 : key = f"{member['payer'].strip().upper()}||{member['plan'].strip().upper()}"
        # Best alternative : use a tuple key
        key = (member["payer"].strip().upper(), member["plan"].strip().upper())
        
        patients_on_same_coverage.setdefault(key, []).append(member["member_id"])
    return [patients for patients in patients_on_same_coverage.values() if len(patients) > 1]

members = [{"member_id": "M1", "payer": "AETNA",  "plan": "PPO-GOLD"},
       {"member_id": "M2", "payer": "aetna",  "plan": "ppo-gold"},
       {"member_id": "M3", "payer": "CIGNA",  "plan": "HMO-BASE"}]

print(f"{find_conflicting_coverage(members)}")
print("we are storing a combination of payer and plan as key, and value as an array of memberId, the space complexity is O(n)")
print("We are iterating over each member only once and hence the time complexity is O(n)")
```
